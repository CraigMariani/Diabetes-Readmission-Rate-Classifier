```python
import pandas as pd
import joblib
```

```python

# Load your processed test or full dataset
X = pd.read_csv("../data/model_ready.csv")  # or whatever your final input CSV is
model = joblib.load("models/rf_model.pkl")

# Predict probabilities
probabilities = model.predict_proba(X)[:, 1]

# Save as CSV
df_probs = pd.DataFrame({"pred_prob": probabilities})
df_probs.to_csv("data/all_probs.csv", index=False)
```
